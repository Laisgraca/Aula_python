#**Escopo de Variáveis**

O escopo de uma variável define onde ela pode ser acessada no código.


*   Variáveis Globais: Declaradas fora de qualquer função, podem ser acessadas de qualquer lugar do código.

*   Variáveis Locais: Declaradas dentro de uma função. Só podem ser acessadas dentro daquela função.



```python
# Variável global
saudacao = 'Olá, seja bem vindo '
nome = ' Aluno Mika'


# Funcao
def minha_funcao():
  # Variável local
  nome ='Acsa'
  print(f'\n Dentro da função : {nome}')
  print(f'\n Acessando a variável global de dentro da função : {saudacao}')

minha_funcao()
print(f'\n Fora da função : {nome}')
print(f'\n  Fora da função : {saudacao}')
```

```python
# Funcao
def minha_funcao():
  #variável local
  nome_local='Acsa'
  print(f'\n Dentro da função : {nome_local}')


minha_funcao()
# Tentar acessar a variável local fora da função resultará em um erro
print(f'\n Tentando acessar a variável nome fora da  da função : {nome_local}')


```
