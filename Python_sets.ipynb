#Estrutura de Dados  - **Sets** (conjuntos)
Conjuntos são coleções não ordenadas de itens únicos e mutáveis. Eles são úteis para remover duplicatas e realizar operações matemáticas de conjuntos (união e interseção)



```python
# Criando um conjunto note que itens duplicados são removidos
conjunto = {1,2,2,3,5,6,6,8,9}
print(f'Conjunto de números sem duplicatas: {conjunto}')
```

```python
type(conjunto)
```

```python
# Adicionando um item
conjunto.add(10)
print(f'Conjunto com ítem adicionado: {conjunto}')
```

```python
# Removendo um item
conjunto.remove(5)
print(f'Removendo um item: {conjunto}')

```

```python
# Operações de conjuntos
conjunto_a = {1,2,6,9}
conjunto_b = {3,4,2,9,6}

# União de todos os elementos de ambos os conjuntos
uniao = conjunto_a.union(conjunto_b)
print(f'União de A e B {uniao}')
```

```python
# interseção dos elementos que estão em ambos os conjuntos
intersecao = conjunto_a.intersection(conjunto_b)
print(f'Interseção de A e B: {intersecao}')
```
