#Entrada de dados com Input
A funçao input sempre retorna uma string por isso, precisa fazer o type casting.


```python

# Pedindo o nome do usuário string

nome = input("Qual é o seu nome? : ")

# pedindo a idade do usuário precisa converter para int
idade_user = int(input("Qual é a sua idade? : "))


from datetime import date
#pega o ano corrente na data definida pelo sistema operacional da sua máquina
ano_atual = date.today().year

#processando os dados
ano_nasc = ano_atual - idade_user

#Exibindo o resultado
print(f" Você tem {idade_user} anos, e nasceu em : {ano_nasc}")
```
