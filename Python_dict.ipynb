# Estrutura de Dados - **Dicionários**
Os dicionários são coleções de pares, chave : valor . São mutáveis

```python
# criando um dicionário
aluno = {
    "Nome":"Mika",
    "Idade": 18,
    "Nota": 8.5,
    "aluno_ativo" : True ,
    "Curso":"Tecnologia da Informação"
}
```

```python
print(f"Dicionário do aluno {aluno}")
```

```python
type(aluno)
```

```python
# Acessando um valor pe chave
print(f"Nome do aluno: {aluno['Nome']} ")
print(f"Curso do aluno: {aluno.get('Curso')}") # .get()forma segura de acessar as chaves
```

```python
# Adicionando um novo par chave-valor
aluno ["Cidade"] = "São Paulo"
print(f"Dicionário atualizado :\n {aluno} ")
```

```python
# Modificando um valor existente
aluno["Idade"] = 22
print(f"Idade atualizada: {aluno['Idade']}")
```

```python
# Apagando par chave-valor
del aluno["Nota"]
print(f"Dicionário atualizado após remover chave nota: \n {aluno}")
```
