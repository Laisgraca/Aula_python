# **Manipulação de Strings**
Strings são sequências de caracteres que possuem muitos métodos úteis

```python
frase = "Eu amo programar em Python" # Define uma variável do tipo string
```

```python
# Concatenação
nome = "Mika"
saudacao = "Olá,Seja Bem vinda " + nome + "!"
print(saudacao)



```

```python
# Tamanho da string(contando com  os espaços)
print(f"tamanho da frase: {len(frase)}")
```

```python
# Maiúsculas e Minúsculas
print(f'Frase em maiúscula {frase.upper()}')
print(f'Frase em minúscula {frase.lower()}')
```

```python
# Remover espaços em branco do início e do fim
frase_sem_espaco = frase.strip()
print(f'Frase sem espaços: {frase_sem_espaco}')
```

```python
# Substituir texto (em tempo de execução com o método replace)
print(f"Substituindo 'amo' por 'gosto de': {frase_sem_espaco.replace('amo','gosto de')}")
```

```python
print(frase_sem_espaco) # veja que o método replace não altera o valor do objeto, somente em tempo de execução.
```

```python
# Fatiamento slicing (Acessando partes de uma string)
# Índice em python começa em 0
print(frase_sem_espaco)
print(f"O primeiro caractere: {frase_sem_espaco [0]}")
print(f"A palavra programar: {frase_sem_espaco [7:16]}") # Do índice 7 ao 15
```
