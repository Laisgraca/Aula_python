# Estrutura de Dados - **Listas**

```python
# Criando uma Lista
frutas = ["Kiwi","Morango","Laranja","Banana"]
print(f'Lista de frutas {frutas}')
```

```python
# Verificar o tipo
type(frutas)
```

```python
#acessando um item pelo índice
print(f"A primeira fruta é {frutas[0]}")
print(f"A última fruta é {frutas[-1]}")
```

```python
# Adicionando um item ao final da lista
frutas.append("Manga")
print(f'Ítem adicionado com sucesso! {frutas}')
```

```python
# utilize o remove para remover um item
frutas.remove("Banana")
print(f"Ítem removido com sucesso! {frutas}")
```

```python
# utilize o pop para remover com base no  índice
frutas.pop(0)
print(f"Ítem removido através do índice {frutas}")
```

```python
# modificando um Ítem
frutas[0] = "Abacaxi"
print(f'Lista após a modificação do primeiro ítem (morango por abacaxi) {frutas}')
```

```python
# verificando o tamanho da lista
print(f"O tamanho da lista é {len(frutas)} frutas")
```

```python
# apagando a lista
del frutas
```
