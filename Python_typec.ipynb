#Conversão Entre tipos de Dados (**type casting**)

```python
#convertendo string para número integer
num_str="25"
num_int=int(num_str)
print(f"'string'{num_str}' para inteiro: {num_int} , tipo:{type(num_int)}")
```

```python
#convertendo string para número float
num_str="3.14"
num_float=float(num_str)
print(f" 'string'{num_str}' para float: {num_float} , tipo: {type(num_float)}")
```

```python
#convertendo número para string
num_int=42
num_str=str(num_int)
print(f" 'int'{num_int}' para string: {num_str} , tipo: {type(num_str)}")
```

```python
#erro ao converter texto para número (O Python verifica se aquele texto realmente representa um número válido.)
texto= "teste"
txt_num = int(texto)
print(f"{txt_num}")

```

```python
#convertendo entre estruturas de dados
lista_duplicatas = [1,2,3,4,4,5,5]
conjunto_unico = set(lista_duplicatas)
lista_sem_duplicatas = list(conjunto_unico)
print(f"Lista com duplicatas: {lista_duplicatas}")
print(f"Conjunto sem duplicatas: {conjunto_unico}")
print(f"Lista sem duplicatas: {lista_sem_duplicatas}")
```
