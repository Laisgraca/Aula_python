# Estrutura de Dados - **Tuplas**

As tuplas são coleções ordenadas e imutáveis de itens. uma vez criadas não podem ser alteradas

```python
# Criando uma tupla
coordenadas = ("10.5","22,6","25,5")
print(coordenadas)
```

```python
type(coordenadas)
```

```python
# Acessando um ítem pelo índice
print(f"coordenada x: {coordenadas[0]}")
print(f"coordenada y: {coordenadas[1]}")
print(f"coordenada z: {coordenadas[2]}")
```

```python
# Tentativa de modificar uma tupla ocorrerá um erro
coordenadas[0]="20.5"
```

```python
# Tuplas são boas para dados que não podem ser alterados como meses do ano
dias_semana=("Segunda","Terça","Quarta","Quinta","Sexta","Sábado","Domingo")
print(f"O primeiro dia da semana é: {dias_semana[0]}")
```
